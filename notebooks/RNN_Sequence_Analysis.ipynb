# RNN-Based Sequence Classifier

## 1. Introduction

Recurrent Neural Networks (RNNs) are neural networks designed to process
sequential data. Unlike traditional neural networks, RNNs maintain a hidden
state that allows information from previous time steps to influence the
processing of later time steps.

In this project, an RNN is developed to classify numerical sequences into
two categories:

- Class 0: Decreasing sequence
- Class 1: Increasing sequence

The project also investigates the effect of different sequence lengths on
classification performance. Three sequence lengths are considered:

- 5
- 10
- 20

After training and evaluation, the model is applied to new sequences that
were not included in the training dataset.

## 5. Sequence Representation

Each numerical sequence is represented as a series of values arranged in
time-step order.

For example, a sequence of length 5 is represented as:

[2.0, 3.1, 4.0, 5.2, 6.1]

An RNN expects the input in the form:

(number of samples, sequence length, number of features)

Since each time step contains one numerical value, the number of features
is 1.

Therefore, for 100 samples with sequence length 5, the input shape is:

(100, 5, 1)

For sequence length 10:

(100, 10, 1)

For sequence length 20:

(100, 20, 1)

The RNN processes the sequence one time step at a time while maintaining a
hidden state that captures information from previous time steps.

```python
import torch

example_sequence = [2.0, 3.1, 4.0, 5.2, 6.1]

x = torch.tensor(
    example_sequence,
    dtype=torch.float32
).view(1, 5, 1)

print("Sequence:")
print(example_sequence)

print("\nRNN input shape:")
print(x.shape)
```

## 6. RNN Architecture

The classifier consists of an RNN layer followed by a fully connected
classification layer.

The architecture is:

Input Sequence
      ↓
RNN Layer
      ↓
Hidden Representation
      ↓
Last Time-Step Output
      ↓
Fully Connected Layer
      ↓
Binary Classification

### RNN Layer

The RNN processes each element of the sequence sequentially. At every
time step, the current input and the hidden state from the previous time
step are used to calculate a new hidden state.

The hidden state allows the network to retain information about the earlier
elements of the sequence.

### Fully Connected Layer

The output from the final time step is passed to a fully connected layer.
This layer produces a single value representing the classification score.

### Activation

During evaluation, the output is passed through the sigmoid function to
obtain a value between 0 and 1.

A value close to 1 indicates an increasing sequence, while a value close
to 0 indicates a decreasing sequence.

The model uses Binary Cross Entropy with Logits Loss during training.
